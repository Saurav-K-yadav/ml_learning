Reshaping, stacking, squeezing and unsqueezing

```python
x=torch.arange(1,10)

x,x.shape
```

```python
x_reshaped=x.reshape(3,3)
x_reshaped
```

```python
# View shares memory with initial variable whereas reshape doesnot

z=x.view(1,9)
z[0][0]=100
z,x
```

```python
# Stacking tensors on top of eachother
x_stacked=torch.stack([x,x,x,x],dim=0)
x_stacked
```

```python
# dim=1 gives column wise for each row together
x_stacked=torch.stack([x,x,x,x],dim=1)
x_stacked
```

```python
# squeeze -- removes all single dimensions
y=torch.arange(0,11)
y.shape
y=y.reshape(1,11)
print(f"before squeeze {y.shape}")
y=y.squeeze()
print(f"before squeeze {y.shape}")
```

```python
# unsqueeze--> adds an extra dimension
z=y.unsqueeze(dim=1)
z.shape,z
```

```python
# Permute --> rearranges dimensions in a specified order
s=torch.rand((2,5,1))
print(s.shape)
s=s.permute((0,1,2)) #takes the position of each dimension
s
```
